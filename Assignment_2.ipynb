```python

 #Question 2



Roll_no = "1024170063"

L = [int(digit) * 10 for digit in Roll_no]

print(" L:", L)



scores = tuple(L[:8])

print("Scores tuple:", scores)



highest = max(scores)
highest_index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)


lowest = min(scores)
lowest_count = scores.count(lowest)

print("Lowest score:", lowest)
print("Num of times lowest score comes:", lowest_count)



reversed_scores = list(reversed(scores))

print("Reversed tuple :", reversed_scores)



score = int(input("Enter a score : "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score is not present in tuple.")



try:
    scores[0] = 100
except TypeError as error:
    print("Error:", error)




first_score, second_score, *remaining_scores = scores

print("Ist score:", first_score)
print("2nd score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
#ques 3


import random


roll_no = 1024170063
random.seed(roll_no)



numbers = []

for i in range(100):
    number = random.randint(100, 900)
    numbers.append(number)

print("Random numbers are :")
print(numbers)



odd_numbers = []

for number in numbers:
    if number % 2 != 0:
        odd_numbers.append(number)

print("\nOdd numbers:")
print(odd_numbers)

print("Num of odd numbers:", len(odd_numbers))



even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print("\nEven numbers:")
print(even_numbers)

print("Num of even numbers:", len(even_numbers))




def is_prime(number):
    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True



prime_numbers = []

for number in numbers:
    if is_prime(number):
        prime_numbers.append(number)

print("\nPrime numbers:")
print(prime_numbers)

print("Num of prime numbers:", len(prime_numbers))



prime_numbers_comprehension = [number for number in numbers if is_prime(number)]

print("Prime numbers using list comprehension:")
print(prime_numbers_comprehension)



most_frequent_number = numbers[0]
maximum_count = numbers.count(numbers[0])

for number in numbers:
    count = numbers.count(number)

    if count > maximum_count:
        maximum_count = count
        most_frequent_number = number

print("\nMost frequently occurring number:", most_frequent_number)
print("Number of times it occurs:", maximum_count)
```

```python
#question 4



roll_no = "1024170063"



digits = [int(digit) for digit in roll_no]

print("Digits:", digits)



A = {digit * 7 for digit in digits}


B = {digit * 9 for digit in digits}

print("\nSet A:", A)
print("Set B:", B)



unionset = A.union(B)

print("\nUnion of a and b:", unionset)



intersection_set = A.intersection(B)

print("Intersection of A and B:", intersection_set)


A_minus_B = A.difference(B)

print("A - B:", A_minus_B)


B_minus_A = B.difference(A)

print("B - A:", B_minus_A)





symmetric_diff = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_diff)




print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))



X = int(input("\nEnter value to remove from Set A: "))


A.discard(X)

print("Set A after discard:", A)

```

```python
#ques 5



my_dict = {
    "name": "Gagandeep Kaur",
    "roll_no": "1024170063",
    "branch": "CSE",
    "age": 19,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)


my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)



my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)



my_dict["age"] = my_dict["age"] + 1

print("\nAfter increasing age by 1:")
print(my_dict)




dict_pop = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("\nAfter deleting branch using pop():")
print(dict_pop)

print("Value returned by pop():", removed_branch)




dict_del = my_dict.copy()

del dict_del["branch"]

print("\nAfter deleting branch using del:")
print(dict_del)



print("\nKey → Value:")

for key, value in my_dict.items():
    print(key, "→", value)




print("\nChecking for email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")




friend_dict = {
    "name": "Sargun",
    "roll_no": "1024170085",
    "branch": "ECE",
    "age": 20,
    "city": "Mohali"
}


merged_dict = {**my_dict, **friend_dict}

print("\nFriend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)



string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)
```
