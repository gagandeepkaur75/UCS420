```python
#Question 1
import numpy as np


temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

print("Original temperatures:", temperature)


# (a)
corrected_temperature = temperature + 2

print("\nTemperatures after adding 2°C:", corrected_temperature)


# (b)
fahrenheit = (9 / 5) * corrected_temperature + 32

print("\nTemperatures in Fahrenheit:", fahrenheit)


# (c)
greater_than_32 = corrected_temperature[corrected_temperature > 32]

print("\nReadings greater than 32°C:", greater_than_32)


# (d)
count = np.sum(corrected_temperature > 32)

print("Number of readings greater than 32°C:", count)
```

```python
#Question 2
import numpy as np


steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:")
print(steps)


# (a)
total_steps = np.sum(steps)

print("\nTotal steps:", total_steps)


# (b)
mean_steps = np.mean(steps)

print("Mean steps:", mean_steps)


# (c)
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)


# (d)
total_each_day = np.sum(steps, axis=0)

print("\nTotal steps for each day:", total_each_day)


# (e)
total_each_user = np.sum(steps, axis=1)

print("Total steps for each user:", total_each_user)


# (f)
max_position = np.argmax(steps)


row, column = np.unravel_index(max_position, steps.shape)

print("\nMaximum value:", steps[row, column])
print("User/Day position of maximum:", (row, column))
```

```python
#Question 3
import numpy as np

# (a)
original = np.array([1, 2, 3, 4, 5, 6])

print("Original array:")
print(original)


# (b)
subset = original[1:5]

print("\nSubset:")
print(subset)


# (c)
subset[0] = 999

print("\nAfter modifying subset:")
print("Original:", original)
print("Subset:", subset)


# d)
copied_array = original[1:5].copy()


copied_array[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)


# (e)
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)


# (f)

# First row
first_row = matrix[0, :]
print("\nFirst row:")
print(first_row)

# Last row
last_row = matrix[-1, :]
print("\nLast row:")
print(last_row)

# Second column
second_column = matrix[:, 1]
print("\nSecond column:")
print(second_column)

# Rows 1-2 and columns 2-3
# Using Python indexing: rows 1 to 2 = 1:3
# columns 2 to 3 = 2:4
selected_elements = matrix[1:3, 2:4]

print("\nRows 1-2 and columns 2-3:")
print(selected_elements)


# (g)

flattened_array = matrix.flatten()
ravel_array = matrix.ravel()

print("\nFlatten using flatten():")
print(flattened_array)

print("\nFlatten using ravel():")
print(ravel_array)


# h)
ravel_array[0] = 1000

print("\nAfter modifying ravel array:")
print("Ravel array:", ravel_array)
print("Original matrix:")
print(matrix)


# i)
flattened_array[0] = 2000

print("\nAfter modifying flatten array:")
print("Flatten array:", flattened_array)
print("Original matrix:")
print(matrix)


# j)

print("\nProperties of original 3 x 4 matrix:")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
#Queston 4
import numpy as np

y = np.array([40, 65, 30, 85])


X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])


# (a)

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)


# b)

X_T = X.T

print("\nTranspose of X:")
print(X_T)


# c)

XTX = X.T @ X

print("\nX.T @ X:")
print(XTX)


# d)

XTX_inverse = np.linalg.inv(XTX)

print("\nInverse of X.T @ X:")
print(XTX_inverse)


# e)

beta = XTX_inverse @ X.T @ y

print("\nOLS coefficients (beta):")
print(beta)


# f)

print("\nCoefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])


# g)

new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\nNew user:", new_user)
print("Predicted assistance score:", predicted_score)
```
