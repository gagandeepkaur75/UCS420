```python
#Queston 1

import numpy as np
import matplotlib.pyplot as plt


M = float(input("Enter the value of M: "))


x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)


plt.plot(x, y1, 'r-', label='y = M*x²')
plt.plot(x, y2, 'b--', label='y = M*sin(x)')


plt.title('Graphs of y = M*x² and y = M*sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)


plt.show()
```

```python
#Question 2

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

subjects = ['English', 'Math', 'Java', 'Machine learning', 'Cloud']
scores = [75, 82, 88, 88, 85]

df = pd.DataFrame({'Subject': subjects, 'Score': scores})

plt.figure(figsize=(8, 5))
ax = sns.barplot(x='Subject', y='Score', data=df, palette='Set2')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)

plt.title('Scores of Five Subjects')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.grid(axis='y')
plt.show()
```

```python
#Question 3

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1024170063)

data = np.random.randn(50)

cumulative_sum = np.cumsum(data)
noise = np.random.randn(50)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(cumulative_sum)
plt.title('Cumulative Sum')
plt.xlabel('Index')
plt.ylabel('Cumulative Sum')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.scatter(range(50), noise)
plt.title('Random Noise')
plt.xlabel('Index')
plt.ylabel('Noise')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(data)
plt.title('Random Data')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.hist(data)
plt.title('Distribution of Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
#Question 4
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

sns.set_style("whitegrid")

plt.figure(figsize=(10, 5))
sns.lineplot(x='month_number', y='total_profit', data=df, marker='o')
plt.title('Total Profit of All Months')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 6))
sns.lineplot(x='month_number', y='facecream', data=df, label='Face Cream')
sns.lineplot(x='month_number', y='facewash', data=df, label='Face Wash')
sns.lineplot(x='month_number', y='toothpaste', data=df, label='Toothpaste')
sns.lineplot(x='month_number', y='bathingsoap', data=df, label='Bathing Soap')
sns.lineplot(x='month_number', y='shampoo', data=df, label='Shampoo')
sns.lineplot(x='month_number', y='moisturizer', data=df, label='Moisturizer')
plt.title('Product Sales Data')
plt.xlabel('Month Number')
plt.ylabel('Sales')
plt.grid(True)
plt.legend()
plt.show()

df.plot(x='month_number', kind='bar', figsize=(14, 7))
plt.title('Bar Chart of All Features')
plt.xlabel('Month Number')
plt.ylabel('Values')
plt.grid(True)
plt.show()
```
