```python
#Question 2(a)

import numpy as np

array = np.array([11, 22, 33, 66, 44,55])
reversed_arr = array[::-1]

print("Original array:", array)
print("Reversed array:", reversed_arr)

#2(b)(i)
import numpy as np

x = np.array([1,2,3,4,5,1,2,1,1,1])

#unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

#most frequent value
most_frequent = values[np.argmax(counts)]

#its indices
indices = np.where(x == most_frequent)[0]

print("Array x:", x)
print("Most frequent value:", most_frequent)
print("Number of times it occurs:", counts[np.argmax(counts)])
print("Indices:", indices)

#2(b)(ii)
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

#unique values and their frequencies
values, counts = np.unique(y, return_counts=True)

#most frequent value
most_frequent = values[np.argmax(counts)]

#indices
indices = np.where(y == most_frequent)[0]

print("Array y:", y)
print("Most frequent value:", most_frequent)
print("Number of times it occurs:", counts[np.argmax(counts)])
print("Indices:", indices)

```

```python
#Question 4
import numpy as np

gagandeep = np.linspace(10, 100, 25)

print("Array:")
print(gagandeep)

# Dimensions of the array
print("\nDimensions:", gagandeep.ndim)

# Shape of the array
print("Shape:", gagandeep.shape)

# Total number of elements
print("Total elements:", gagandeep.size)

# Data type of each element
print("Data type:", gagandeep.dtype)

# Total number of bytes consumed
print("Total bytes:", gagandeep.nbytes)

# Transpose using reshape()
transpose_gagandeep = gagandeep.reshape(25, 1)

print("\nTranspose using reshape():")
print(transpose_gagandeep)

# Transpose using T attribute
transpose_using_T = gagandeep.T

print("\nTranspose using T:")
print(transpose_using_T)
```

```python
#Question 5

import numpy as np

ucs420_gagandeep = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_gagandeep)


#  mean
print("\nMean:", np.mean(ucs420_gagandeep))

#  median
print("Median:", np.median(ucs420_gagandeep))

# maximum
print("Maximum:", np.max(ucs420_gagandeep))

# minimum
print("Minimum:", np.min(ucs420_gagandeep))

# unique elements
print("Unique elements:", np.unique(ucs420_gagandeep))


# Reshape to 4 rows and 3 columns
reshaped_ucs420_gagandeep = ucs420_gagandeep.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_gagandeep)


# Resize to 2 rows and 3 columns
resized_ucs420_gagandeep = np.resize(ucs420_gagandeep, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_gagandeep)
```
