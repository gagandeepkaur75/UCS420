```python
# Assignment 5.3
n = int(input("Enter n: "))
sum = 0
for num in range(2, n + 1):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        sum = sum + num

print("Sum of prime numbers =", sum)
```

```python
# Assignment 6.1
def add_odd(n):
    sum = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            sum = sum + i

    return sum


n = int(input("Enter n: "))
result = add_odd(n)
print("Sum of odd numbers =", result)
```

```python
# Assignment 6.2
def add_prime(n):
    sum = 0
    for num in range(2, n + 1):

        count = 0

        for i in range(1, num + 1):
            if num % i == 0:
                count = count + 1

        if count == 2:
            sum = sum + num

    return sum


n = int(input("Enter n: "))
result = add_prime(n)
print("Sum of prime numbers =", result)
```
